```python
from pathlib import Path
import sys

project_root = Path.cwd().resolve().parent
if str(project_root) not in sys.path:
    sys.path.insert(0, str(project_root))

dbutils.widgets.text("output_dir", "data")
dbutils.widgets.text("gsm8k_val_size", "500")
dbutils.widgets.text("xsum_train_size", "20000")
dbutils.widgets.text("xsum_val_size", "1000")
dbutils.widgets.text("xsum_test_size", "1000")
dbutils.widgets.text("seed", "42")
```

```python
from src.data.prep.gsm8k import prepare_gsm8k
from src.data.prep.xsum import prepare_xsum

output_dir = project_root / dbutils.widgets.get("output_dir")
seed = int(dbutils.widgets.get("seed"))

gsm8k_manifest = prepare_gsm8k(
    output_dir=output_dir,
    manifest_path=output_dir / "manifests" / "gsm8k.json",
    val_size=int(dbutils.widgets.get("gsm8k_val_size")),
    seed=seed,
)
gsm8k_manifest["counts"]
```

```python
xsum_manifest = prepare_xsum(
    output_dir=output_dir,
    manifest_path=output_dir / "manifests" / "xsum.json",
    train_size=int(dbutils.widgets.get("xsum_train_size")),
    val_size=int(dbutils.widgets.get("xsum_val_size")),
    test_size=int(dbutils.widgets.get("xsum_test_size")),
    seed=seed,
)
xsum_manifest["counts"]
```
